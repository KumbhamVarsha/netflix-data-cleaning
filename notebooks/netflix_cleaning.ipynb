```python
import pandas as pd

print("Pandas imported successfully!")
```

```python
df = pd.read_csv("../Data/netflix_titles.csv")

print("Dataset loaded successfully!")
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage.round(2))
```

```python
df["director"] = df["director"].fillna("Not Available")
df["cast"] = df["cast"].fillna("Not Available")
df["country"] = df["country"].fillna("Not Available")
```

```python
df[["director", "cast", "country"]].isnull().sum()
```

```python
df["rating"] = df["rating"].fillna("Not Rated")
```

```python
df["rating"].isnull().sum()
```

```python
df[df["duration"].isnull()]
```

```python
missing_duration = df["duration"].isnull()

df.loc[missing_duration, "duration"] = df.loc[missing_duration, "rating"]

df.loc[missing_duration, "rating"] = "Not Rated"
```

```python
df.loc[[5541, 5794, 5813], ["title", "rating", "duration"]]
```

```python
df[df["date_added"].isnull()]
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
```

```python
df["date_added"].dtype
```

```python
df["date_added"].isnull().sum()
```

```python
original_dates = pd.read_csv("../Data/netflix_titles.csv", usecols=["date_added"])

print("Missing dates in original CSV:")
print(original_dates["date_added"].isnull().sum())
```

```python
df["date_added"] = pd.to_datetime(
    original_dates["date_added"],
    format="mixed",
    errors="coerce"
)
```

```python
df["date_added"].isnull().sum()
```

```python
df["duration"].value_counts().head(20)
```

```python
df["duration_value"] = df["duration"].str.extract(r"(\d+)")

df["duration_unit"] = df["duration"].str.extract(r"(min|Seasons?)")

df[["type", "title", "duration", "duration_value", "duration_unit"]].head(10)
```

```python
df["duration_value"] = pd.to_numeric(df["duration_value"])
```

```python
df["duration_value"].dtype
```

```python
df.to_csv("../Data/netflix_titles_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
